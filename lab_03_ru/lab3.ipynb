1. Найдите в `viterbi_trellis.svg` две ячейки «кадр/состояние» с низкой стоимостью и одну с высокой. Что кодирует тепловая карта?

Пример низкой стоимости: третий фрейм x /y/. Пример высокой – первый фрейм и /s/. епловая карта кодирует стоимость эмиссии, то есть насколько хорошо каждый кадр соответствует каждому состоянию

2. Проследите путь, выделенный зелёным цветом. Покажите один переход с сохранением состояния и один переход в следующее состояние.

Один переход с сохранением состояния – из f2 в f3, а вот f3 в f4 с изменением

3. Сравните стоимость лучшего пути со стоимостью альтернативного, сославшись на соответствующие строки в `alignment_path.csv` или данные в `summary.json`.

Лучший путь стоит 3.0: sil_start -> /y/ -> /y/ -> /e/ -> /s/ -> sil_end

Следующий лучший путь стоит 3.7: sil_start -> /y/ -> /e/ -> /e/ -> /s/ -> sil_end

4. Для `λ=0` и `λ=0.5` укажите гипотезу-кандидата на первом месте и его акустическую, языковую и объединённую логарифмические оценки из `candidate_scores.csv`.

Для лямбда 0 на первом месте we drink strong coffee, -1.000000,-2.654739,-1.000000 – акустическая, языковая и объединенная оценки соответственно

Для лямбда 0.5 на первом месте we drink green tea, -1.200000,-1.911720,-2.155860 – акустическая, языковая и объединенная оценки соответственно

5. Почему кандидат на первом месте меняется (или не меняется) с изменением λ?

Соотношение участия акустики и языковой модели. Чистая акустика дает we drink strong coffee, модель с большим участием языковой модели дает we drink green tea
